```python
order_amount=[1000,1400,None,"invalid",300.50,400.60]
sum_r=0
for x in order_amount:
    if type(x) ==int or type(x) == float:
        sum_r=sum_r+x
    else:
        continue
print(sum_r)
```

```python
order_amount_while=[1000,1400,None,"invalid",300.50,400.60]
i=0
sum_while=0

while i <len(order_amount_while):
    if type(order_amount_while[i])==int or type(order_amount_while[i])==float:
        sum_while=sum_while+order_amount_while[i]
    else:
        i=i+1
        continue
    i=i+1
print(sum_while)
```

```python
order_amount_while_t=[1000,1400,None,"invalid",300.50,400.60]
i=0
sum_while_true=0

while True:
    if type(order_amount_while_t[i])==int or type(order_amount_while_t[i])==float:
        sum_while_true=sum_while_true+order_amount_while_t[i]
    else:
        i=i+1
        continue
    i=i+1
    if i ==len(order_amount_while_t):
        break
print(sum_while_true)
```

```python
customer_ids =[102,105,102,103,107,109,110,109]

unique_customer=[]

for i in customer_ids:
    if i in unique_customer:
        continue
    else:
        unique_customer.append(i)

print(unique_customer)
```

```python
#create a new list after performing some transformation 

order_amounts=[100,200,50,500,400,900,1200,70]
order_amount_GST=[]

for x in order_amounts:
    order_amount_GST.append(x+x*0.18)
print(order_amount_GST)
```

```python
order_gst =[x+x*.18 for x in order_amounts]
print(order_gst)
```

```python
diff_or_amount=[(100,5),(200,18),(50,12),(500,18),(400,12),(900,5),(1200,18),(70,12)]
include_gst=[]
for x in diff_or_amount:
    include_gst.append(x[0]+x[0]*x[1]/100)
print(include_gst)
```

```python
#list comprehension 
included_gst=[(x[0],x[1],x[0]+x[0]*x[1]/100) for x in diff_or_amount]
print(included_gst)
```

```python
#[[1,1,1],[2,4,8],[3,,9,27]]
sq_list=[]
for i in range (1,4):
    sq_list.append([i,i**2,i**3])

print(sq_list)
```

```python
#count the number of occurance in each word 

input_list=["hello","HELLO","I","AM","AM","Sumit","sumit"]
input_set=set(input_list)
print(input_set)
```

```python
input_lower=[word.lower() for word in input_list]
print(input_lower)
```

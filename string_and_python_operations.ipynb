```python
# Addition
a = 10 + 3
print(a)  # 13

# Subtraction
b = 10 - 3
print(b)  # 7

# Multiplication
c = 10 * 3
print(c)  # 30

# Division (always returns float)
d = 10 / 3
print(d)  # 3.3333333333333335

# Floor Division (returns whole number, discards decimal)
e = 10 // 3
print(e)  # 3

# Modulus (remainder after division)
f = 10 % 3
print(f)  # 1

# Exponentiation (power)
g = 10 ** 3
print(g)  # 1000
```

```python
##Operator Precedence (BODMAS applies)
pythonresult = 10 + 3 * 2   # 16, not 26 (multiplication happens first)
result2 = (10 + 3) * 2  # 26 (brackets first)
print(pythonresult)
print(result2)
```

```python
import math
total_sales= -20000.60
print((math.ceil(total_sales)))
print(math.fabs(total_sales))
print(math.floor(total_sales))
"""
math.ceil() and math.floor() return an integer
math.fabs() always returns a float, unlike built-in abs() which returns int or float depending on input
"""
```

```python

text = "Python"

print(text[0])     # P  (first character) indexing 
print(text[-1])    # n  (last character)
print(text[0:4])   # Pyth  (slicing: start to end-1)
print(text[:3])    # Pyt
print(text[3:])    # hon
print(text[::-1])  # nohtyP  (reversed string)
```

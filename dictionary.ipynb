```python
customers_raw_data="""customer_id,customer_fname,customer_lname,address,city,state,pincode
11599,Mary,Malone,8708 Indian Horse Highway,Hickory,NC,28601
256,David,Rodriguez,7605 Tawny Horse Falls,Chicago,IL,60625
12111,Amber,Franco,8766 Clear Prairie Line,Santa Cruz,CA,95060
8827,Brian,Wilson,8396 High Corners,San Antonio,TX,78240
11318,Mary,Henry,3047 Silent Embers Maze,Caguas,PR,00725"""
```

```python
print(customers_raw_data)
```

```python
customer_header=customers_raw_data.split("\n")[0].split(",")
print(customer_header)
```

```python
customer_data=customers_raw_data.split("\n")[1:]
print(customer_data)
```

```python
customer_dictonary={}
for x in customer_data:
    customer_dictonary[x.split(",")[0]]=tuple(x.split(",")[1:])
print(customer_dictonary)
```

```python
customer_dictonary['11599']
```

```python
cd={x.split(",")[0]:tuple(x.split(",")[1:]) for x in customer_data}
print(cd)
```

```python
nested_dictionary={}
for key,value in cd.items():
    nested_dictionary[key]={
    customer_header[1]:value[0],
    "customer_lname":value[1],
    "address":value[2],
    "city":value[3],
    "state":value[4],
    "pincode":value[5]
    } 
print(nested_dictionary)
```

```python
print(nested_dictionary.get("11318").get("pincode"))
```

```python
raw_orders_data = """1,'2023-07-25 00:00:00.0',11599,CLOSED
2,'2023-07-25 00:00:00.0',256,PENDING_PAYMENT
3,'2023-07-25 00:00:00.0',12111,COMPLETE
4,'2023-07-25 00:00:00.o',12111,CLOSED
5,'2023-07-25 00:00:00.o',256,COMPLETE"""

orders_data = raw_orders_data.split("\n")

combined_result = []
for x in orders_data:
    order = x.split(",")
    if len(order) < 4:
        print(f"Skipping malformed row: {order}")
        continue
    combined_result.append((order[0], order[1], order[2], order[3], customer_dictonary.get(order[2])))

print(combined_result)
```
